## 数据解析转换

```python
file = "./datas/word_segment/original/01.txt"
k = 0
with open(file, "r", encoding="utf-8") as reader:
    for line in reader:
        print(line)
        k += 1
        print(line.split(" "))
        if k > 2:
            break
```

```python
in_dir = "./datas/word_segment/original/"
in_names = os.listdir(in_dir)
datas = []
for name in in_names:
    in_file = os.path.join(in_dir, name)
    if not os.path.isfile(in_file):
        continue

    # 开始加载文件内容
    with open(in_file, "r", encoding="utf-8") as reader:
        for line in reader:
            original_text = ""
            entities = []
            line = line.strip()
            if line:
                tokens = line.split(" ")[1:]
                for token in tokens:
                    token = token.strip()
                    if token:
                        try:
                            word, _ = token.split("/")
                            start_pos = len(original_text)
                            original_text = original_text + word
                            end_pos = len(original_text)
                            entities.append({
                                "label_type": "Word",
                                "overlap": 0,
                                "start_pos": start_pos,
                                "end_pos": end_pos
                            })
                        except Exception as e:
                            print(f"解析异常 {line} {token} {e}")
                            entities = []
                            break
            if len(entities) > 0 and (entities[0]['start_pos'] == 0) and (entities[-1]['end_pos'] == len(original_text)):
                datas.append({
                    "originalText": original_text,
                    "entities": entities
                })
```

```python
train_data_file = "./datas/word_segment/train.jsonl"
test_data_file = "./datas/word_segment/test.jsonl"

# 输出文件夹创建
os.makedirs(os.path.dirname(train_data_file), exist_ok=True)
os.makedirs(os.path.dirname(test_data_file), exist_ok=True)

# 加载所有数据, 并按照随机分配给训练集和测试集
train_datas, test_datas = [], []
test_size = 0.05
for obj in datas:
    if random.random() < test_size:
        test_datas.append(obj)
    else:
        train_datas.append(obj)

# 数据输出
with jsonlines.open(train_data_file, "w") as writer:
    writer.write_all(train_datas)
print(f"输出训练数据完成:{train_data_file} - {len(train_datas)}")
with jsonlines.open(test_data_file, "w") as writer:
    writer.write_all(test_datas)
print(f"输出测试数据完成:{test_data_file} - {len(test_datas)}")
```

## 训练模型
- 更改一下长文本划分的逻辑；

```python
from ner.trainer.trainer import Trainer
from ner.config import Config
from ner.datas.tokenizer import Tokenizer

bert_path = "google-bert/bert-base-chinese"

tokenizer = Tokenizer(
    vocabs=r'./datas/vocab.txt'
)

data_root_dir = r"./datas/word_segment"
cfg = Config(
    output_dir="./output/word_segment/",
    tokenizer=tokenizer,  # 分词器
    label2id=os.path.join(data_root_dir, "label2id.json"),  # 类别标签名称到id的映射

    train_file=os.path.join(data_root_dir, "train.jsonl"),  # 训练数据对应文件
    eval_file=os.path.join(data_root_dir, "test.jsonl"),  # 模型评估数据对应文件

    #total_epoch=100, # 第一次训练时候的参数
    total_epoch=10,
    batch_size=32,  # 批次大小
    lr=0.01,  # 模型训练学习率

    # network_type='lstm',  # 网络类型 可选lstm、bert
    network_type='bert',  # 网络类型 可选lstm、bert
    lstm_layers=3,  # LSTM的层数
    lstm_hidden_size=768,
    bert_path=bert_path,  # Bert模型迁移路径
    max_length=512,
    #freeze=True,  # 给定迁移模型的时候冻结参数 第一次训练时候的参数
    freeze=False,  # 给定迁移模型的时候冻结参数
    device="cuda",
    max_no_improved_epoch=20
)
trainer = Trainer(config=cfg)
```

## 模型调用
- PS: 首先运行 server_starter.py 文件启动web接口服务

```python
import requests

url = "http://127.0.0.1:9001/predict"

def tt_get(text=None):
    if text is None:
        text = "帮我查一下最近去湘潭的高铁什么时候出发"

    response = requests.get(
        url, json={'text': text}
    )
    if response.status_code == 200:
        print("访问服务器成功GET")
        # 明确的知道服务器返回的是json格式，那么可以直接调用json()这个方法转换为字典对象
        result = response.json()
        print(result)
        print(type(result))
        if result['code'] == 0:
            print(f"调用模型成功GET，结果为:{result['data']}")
            print(f"调用模型成功GET，结果为:{[a['entity'] for a in result['data']]}")
        else:
            print(f"调用模型服务器处理异常，异常信息为:{result['msg']}")
    else:
        print("访问服务器失败：网络异常等等")

def tt_post(text=None):
    # NOTE: 在fastapi框架中，post请求的参数必须通过json参数给定
    if text is None:
        text = "帮我查一下最近去湘潭的高铁什么时候出发"
        text = "去虹桥火车站最快的路线是什么"
        text = "从我家到嘉兴北站怎么走"
        text = "附近有什么好玩的游乐场吗"
    response = requests.post(
        url,
        json={'text': text},
    )
    if response.status_code == 200:
        print("访问服务器成功POST")
        # 明确的知道服务器返回的是json格式，那么可以直接调用json()这个方法转换为字典对象
        result = response.json()
        print(result)
        print(type(result))
        if result['code'] == 0:
            print(f"调用模型成功POST，结果为:{result['data']}")
            print(f"调用模型成功GET，结果为:{[a['entity'] for a in result['data']]}")
        else:
            print(f"调用模型服务器处理异常，异常信息为:{result['msg']}")
    else:
        print("访问服务器失败：网络异常等等")
```

```python

tt_get()
tt_post()
```
